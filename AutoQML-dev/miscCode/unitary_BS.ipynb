```python
def BS_U(thetas,phis,connections,wires):
    # Running overall unitary
    U = np.eye(wires, dtype=complex)

    # For each BS
    for idx, bs_wires in enumerate(connections):

        theta = thetas[idx]
        phi = phis[idx]
        c, s = np.cos(theta), np.sin(theta)

        # Blank matrix in full space
        BS_embed = np.eye(wires, dtype=complex)

        # Embed BS into full space
        BS_embed[bs_wires[0],bs_wires[0]] = c
        BS_embed[bs_wires[0],bs_wires[1]] = -np.exp(-1j * phi) * s
        BS_embed[bs_wires[1],bs_wires[0]] = np.exp(1j * phi) * s
        BS_embed[bs_wires[1],bs_wires[1]] = c

        # Add to overall unitary
        U = BS_embed @ U
    
    # Plot / print
    return U
```
